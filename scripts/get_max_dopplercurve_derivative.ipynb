```python
import csv
data = []
with open('../HULIANWANG_DIGUI-54_20260713_005129_doppler_data.csv', mode='r') as file:
    reader = csv.reader(file)
    next(reader)
    for row in reader:
        data.append([float(x) for x in row])
```

```python
dataderivative = []
for i in range(1, len(data)-1):
    dataderivative.append([data[i][0], data[i+1][1] - data[i][1]])

print(dataderivative)
```

```python
from matplotlib import pyplot as plt

plt.plot([x[0] for x in dataderivative], [x[1] for x in dataderivative])
plt.xlabel("time (sec)")
plt.ylabel("1st derivative of the doppler curve")
plt.show()
```
