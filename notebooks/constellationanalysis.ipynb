```python
from collections import Counter

with open('../activesatellites.txt', 'r') as f:
    active_satellites = f.readlines()
```

```python
substrings = []
for name in active_satellites:
    substring = ''
    for char in name:
        if char.isalpha():
            substring += char
        else:
            break
    if len(substring) > 2:
        substrings.append(substring.lower())

most_common = Counter(substrings).most_common(30)
print(most_common)
```

```python
from matplotlib import pyplot as plt
labels, counts = zip(*most_common[2:])
plt.bar(labels, counts)
plt.xticks(rotation=90)
plt.show()
```
